```python

from pathlib import Path
import json
from pprint import pprint

path = Path("./logs/mgwr-2025-10-22-23-36-52-log/model_info.json")
with path.open("r", encoding="utf-8") as f:
    mode_info = json.load(f)
bandwidth_vectors = mode_info['bandwidth_optimization']
```

```python
import ast

rewards = []
bandwidths = []

for v in bandwidth_vectors:
    v["bandwidth"] = ast.literal_eval(v["bandwidth"]) if isinstance(v["bandwidth"], str) else v["bandwidth"]
    bandwidths.append(v["bandwidth"])
    rewards.append(v["reward"])

import numpy as np
import pandas as pd

data = pd.DataFrame({
    "bandwidth_0": [b[0] for b in bandwidths],
    "bandwidth_1": [b[1] for b in bandwidths],
    "bandwidth_2": [b[2] for b in bandwidths],
    "reward": rewards
})

# sort ascending by reward and then by the bandwidth columns
data = data.sort_values(by=["reward"], ascending=True).reset_index(drop=True)
data
```
